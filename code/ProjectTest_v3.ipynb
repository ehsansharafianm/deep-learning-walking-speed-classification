```python
import pandas as pd
import numpy as np
```

```python
# Define the file paths
file_paths = [
    '/content/ProjectData/GP1_0.6_marker.csv',
    '/content/ProjectData/GP1_0.7_marker.csv',
    '/content/ProjectData/GP1_0.8_marker.csv',
    '/content/ProjectData/GP1_0.9_marker.csv',
    '/content/ProjectData/GP1_1.0_marker.csv',
    '/content/ProjectData/GP1_1.1_marker.csv',
    '/content/ProjectData/GP1_1.2_marker.csv',
    '/content/ProjectData/GP1_1.3_marker.csv',
    '/content/ProjectData/GP1_1.4_marker.csv',
    '/content/ProjectData/GP1_1.5_marker.csv',
    '/content/ProjectData/GP1_1.6_marker.csv',
    '/content/ProjectData/GP1_1.7_marker.csv'
]

# Initialize an empty DataFrame
data = pd.DataFrame()

# Create DataFrames for each secim value and concatenate with marker data
for i, file_path in enumerate(file_paths, start=6):
    # Read marker data
    marker_data = pd.read_csv(file_path)

    # Create DataFrame with 'secim' values
    secim_data = pd.DataFrame({'secim': i * np.ones(len(marker_data))})

    # Concatenate marker and secim DataFrames
    combined_data = pd.concat([marker_data, secim_data], axis=1)

    # Concatenate with the main DataFrame 'data'
    data = pd.concat([data, combined_data], axis=0, ignore_index=True)

data.head()
```

```python
x = data.drop(['secim'], axis = 1)
y = data.secim.values
x.head()
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x , y, test_size= 0.3, random_state= 1)
```

SVM

```python
from sklearn.svm import SVC

svm = SVC(random_state=1)
svm.fit(x_train, y_train)
print("accuracy of svm is :", svm.score(x_test, y_test))
```

Naive Bayes Classifier

```python
from sklearn.naive_bayes import GaussianNB

nb = GaussianNB()
nb.fit(x_train, y_train)
print("accuracy of Naive Bayes is :", nb.score(x_test, y_test))

```

Decision Tree Classifier

```python
from sklearn.tree import DecisionTreeClassifier

dt =DecisionTreeClassifier()
dt.fit(x_train, y_train)
print("Accuracy of Decision Tree is :", nb.score(x_test, y_test))
```

Random Forest Classifier


```python
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(n_estimators = 100, random_state = 1)
rf.fit(x_train, y_train)

print('Accuracy of Random Forest is :', rf.score(x_test, y_test))
```

Logistic Regression Classifier

```python
from sklearn.linear_model import LogisticRegression
lr = LogisticRegression()
lr.fit(x_train, y_train)

print("Accuracy of Logistic Regression is :'", lr.score(x_test, y_test))
```

KNN with K = 3

```python
from sklearn.neighbors import KNeighborsClassifier
knn = KNeighborsClassifier(n_neighbors= 3)
knn.fit(x_train, y_train)
prediction = knn.predict(x_test)

print('K = {} nn score: {}'.format(3, knn.score(x_test, y_test)))
```
